```python
# COMPARAISON DE TROIS MLP
# 1) MLP sans régularisation
# 2) MLP avec régularisation L2
# 3) MLP avec régularisation personnalisée par distribution
#    (sparsité KL-divergence + anti-saturation par écart-type)


# IMPORTATIONS

import os
import random
import warnings

import numpy as np
import pandas as pd
import tensorflow as tf
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

warnings.filterwarnings("ignore")

print("="*60)
print("TensorFlow Version :", tf.__version__)

# Vérification GPU (utile sur Kaggle)
gpus = tf.config.list_physical_devices('GPU')

if len(gpus) > 0:
    print("GPU détecté :", gpus[0].name)
else:
    print("Aucun GPU détecté - utilisation du CPU")

print("="*60)

# REPRODUCTIBILITE


SEED = 42

os.environ["PYTHONHASHSEED"] = str(SEED)

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

# HYPERPARAMETRES

LEARNING_RATE = 0.001

EPOCHS = 100

BATCH_SIZE = 32

# Régularisation L2
LAMBDA_L2 = 1e-4

# Régularisation personnalisée - Sparsité (KL-divergence)
LAMBDA_SPARSITY = 0.01      # poids de la pénalité de sparsité
RHO = 0.30                  # taux d'activation moyen cible par neurone

# Régularisation personnalisée - Anti-saturation (écart-type)
LAMBDA_STD = 0.001          # poids de la pénalité anti-saturation
STD_MIN = 0.05              # écart-type minimal toléré par neurone

# CHARGEMENT DU DATASET

data = load_breast_cancer()

X = data.data.astype(np.float32)

y = data.target.astype(np.int32)

print("\nDataset chargé avec succès")
print("-"*60)

print("Nom :", "Breast Cancer Wisconsin")

print("Nombre d'exemples :", X.shape[0])

print("Nombre de caractéristiques :", X.shape[1])

print("Nombre de classes :", len(np.unique(y)))

print("-"*60)

# DIVISION TRAIN / TEST

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=SEED,
    stratify=y
)

# NORMALISATION DES DONNEES

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)

print("\nAprès séparation :")
print("Train :", X_train.shape)
print("Test  :", X_test.shape)

# FONCTION D'EVALUATION

def evaluate_model(model, history, model_name):

    print("\n")
    print("="*60)
    print(model_name)
    print("="*60)

    # Prédictions
    y_prob = model.predict(X_test, verbose=0)

    y_pred = np.argmax(y_prob, axis=1)

    # Accuracy
    accuracy = accuracy_score(y_test, y_pred)

    print(f"Accuracy : {accuracy:.4f}")

    print("\nClassification Report\n")

    print(classification_report(y_test, y_pred))

    print("Confusion Matrix\n")

    print(confusion_matrix(y_test, y_pred))

    return accuracy

# AFFICHAGE DES COURBES

def plot_history(history, title):

    plt.figure(figsize=(12,5))

    # Accuracy
    plt.subplot(1,2,1)

    plt.plot(history.history["accuracy"], label="Train")

    plt.plot(history.history["val_accuracy"], label="Validation")

    plt.title(title + " - Accuracy")

    plt.xlabel("Epoch")

    plt.ylabel("Accuracy")

    plt.legend()

    # Loss
    plt.subplot(1,2,2)

    plt.plot(history.history["loss"], label="Train")

    plt.plot(history.history["val_loss"], label="Validation")

    plt.title(title + " - Loss")

    plt.xlabel("Epoch")

    plt.ylabel("Loss")

    plt.legend()

    plt.tight_layout()

    plt.show()

# STOCKAGE DES RESULTATS

results = []

print("\nToutes les données sont prêtes.")
print("Passer maintenant à la Partie 2 (définition des modèles).")


# PARTIE 2 : DEFINITION DES TROIS MODELES

from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Dense, Layer
from tensorflow.keras.regularizers import l2
from tensorflow.keras.optimizers import Adam


# COUCHE DE REGULARISATION PERSONNALISEE PAR DISTRIBUTION

class ActivationRegularization(Layer):
    def __init__(self, lambda_sparsity=0.01, rho=0.3,
                 lambda_std=0.001, std_min=0.05,
                 epsilon=1e-8, **kwargs):
        super().__init__(**kwargs)
        self.lambda_sparsity = lambda_sparsity
        self.rho = rho
        self.lambda_std = lambda_std
        self.std_min = std_min
        self.epsilon = epsilon

    def call(self, inputs):

        # ---- Taux d'activation moyen par neurone (sur le batch) ----
        rho_hat = tf.reduce_mean(inputs, axis=0)

        # Clip pour éviter log(0) ou division par 0
        rho_hat = tf.clip_by_value(
            rho_hat,
            self.epsilon,
            1.0 - self.epsilon
        )

        # ---- Pénalité de sparsité (KL-divergence) ----
        kl = (
            self.rho * tf.math.log(self.rho / rho_hat)
            + (1.0 - self.rho) * tf.math.log(
                (1.0 - self.rho) / (1.0 - rho_hat)
            )
        )

        sparsity_penalty = tf.reduce_sum(kl)

        # ---- Ecart-type d'activation par neurone (sur le batch) ----
        std_j = tf.math.reduce_std(inputs, axis=0)

        # Pénalise seulement les neurones trop "figés"
        std_penalty = tf.reduce_sum(
            tf.square(tf.maximum(0.0, self.std_min - std_j))
        )

        # ---- Perte totale ----
        total_loss = (
            self.lambda_sparsity * sparsity_penalty
            + self.lambda_std * std_penalty
        )

        self.add_loss(total_loss)

        # Les activations continuent normalement, inchangées
        return inputs

    def get_config(self):
        config = super().get_config()
        config.update({
            "lambda_sparsity": self.lambda_sparsity,
            "rho": self.rho,
            "lambda_std": self.lambda_std,
            "std_min": self.std_min,
            "epsilon": self.epsilon,
        })
        return config


# MODELE 1 : BASELINE (AUCUNE REGULARISATION)

def build_baseline_model():

    inputs = Input(shape=(X_train.shape[1],))

    x = Dense(
        64,
        activation="sigmoid"
    )(inputs)

    x = Dense(
        32,
        activation="sigmoid"
    )(x)

    outputs = Dense(
        2,
        activation="softmax"
    )(x)

    model = Model(inputs, outputs)

    model.compile(
        optimizer=Adam(learning_rate=LEARNING_RATE),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# MODELE 2 : REGULARISATION L2

def build_l2_model():

    inputs = Input(shape=(X_train.shape[1],))

    x = Dense(
        64,
        activation="sigmoid",
        kernel_regularizer=l2(LAMBDA_L2)
    )(inputs)

    x = Dense(
        32,
        activation="sigmoid",
        kernel_regularizer=l2(LAMBDA_L2)
    )(x)

    outputs = Dense(
        2,
        activation="softmax"
    )(x)

    model = Model(inputs, outputs)

    model.compile(
        optimizer=Adam(learning_rate=LEARNING_RATE),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


# MODELE 3 : REGULARISATION PERSONNALISEE PAR DISTRIBUTION

def build_custom_model():

    inputs = Input(shape=(X_train.shape[1],))

    x = Dense(
        64,
        activation="sigmoid"
    )(inputs)

    # Régularisation par distribution : sparsité (KL) + anti-saturation (std)
    x = ActivationRegularization(
        lambda_sparsity=LAMBDA_SPARSITY,
        rho=RHO,
        lambda_std=LAMBDA_STD,
        std_min=STD_MIN
    )(x)

    x = Dense(
        32,
        activation="sigmoid"
    )(x)

    outputs = Dense(
        2,
        activation="softmax"
    )(x)

    model = Model(inputs, outputs)

    model.compile(
        optimizer=Adam(learning_rate=LEARNING_RATE),
        loss="sparse_categorical_crossentropy",
        metrics=["accuracy"]
    )

    return model


 
# CREATION DES TROIS MODELES
 

baseline_model = build_baseline_model()

l2_model = build_l2_model()

custom_model = build_custom_model()


print("\nRésumé du modèle Baseline")
baseline_model.summary()

print("\nRésumé du modèle L2")
l2_model.summary()

print("\nRésumé du modèle Régularisation personnalisée (distribution)")
custom_model.summary()


 
# PARTIE 3 : ENTRAINEMENT ET COMPARAISON
 


 
# FONCTION D'ENTRAINEMENT
 

def train_model(model, model_name):

    print("\n")
    print("="*60)
    print("ENTRAINEMENT :", model_name)
    print("="*60)

    history = model.fit(
        X_train,
        y_train,
        validation_split=0.2,
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        verbose=1
    )

    return history


 
# ENTRAINEMENT DES TROIS MODELES
 

history_baseline = train_model(
    baseline_model,
    "MLP sans régularisation"
)


history_l2 = train_model(
    l2_model,
    "MLP avec régularisation L2"
)


history_custom = train_model(
    custom_model,
    "MLP avec régularisation par distribution (KL + std)"
)


 
# EVALUATION DES MODELES
 

acc_baseline = evaluate_model(
    baseline_model,
    history_baseline,
    "RESULTAT : MLP Baseline"
)


acc_l2 = evaluate_model(
    l2_model,
    history_l2,
    "RESULTAT : MLP L2"
)


acc_custom = evaluate_model(
    custom_model,
    history_custom,
    "RESULTAT : MLP Régularisation par distribution"
)


 
# STOCKAGE DES RESULTATS
 

results.append(
    {
        "Model": "Baseline",
        "Accuracy": acc_baseline
    }
)


results.append(
    {
        "Model": "L2",
        "Accuracy": acc_l2
    }
)


results.append(
    {
        "Model": "Distribution (KL + std)",
        "Accuracy": acc_custom
    }
)


 
# TABLEAU COMPARATIF FINAL
 

results_df = pd.DataFrame(results)


print("\n")
print("="*60)
print("COMPARAISON FINALE")
print("="*60)

print(results_df)


 
# COURBES D'APPRENTISSAGE
 

plot_history(
    history_baseline,
    "MLP Baseline"
)


plot_history(
    history_l2,
    "MLP L2"
)


plot_history(
    history_custom,
    "MLP Régularisation par distribution"
)


 
# COMPARAISON VISUELLE DES ACCURACY
 

plt.figure(figsize=(8,5))

plt.bar(
    results_df["Model"],
    results_df["Accuracy"]
)

plt.title(
    "Comparaison des performances des trois MLP"
)

plt.ylabel(
    "Accuracy"
)

plt.xticks(
    rotation=45
)

plt.ylim(
    0,
    1
)

plt.grid(
    axis="y"
)

plt.tight_layout()

plt.show()


 
# ANALYSE DES ACTIVATIONS DU MODELE CUSTOM (OPTIONNEL)
 
# Permet de vérifier visuellement l'effet de la régularisation :
# on extrait les activations de la couche régularisée et on
# regarde leur distribution (moyenne et écart-type par neurone).

def analyze_custom_activations(model, X_sample):

    # On récupère la sortie juste après la couche ActivationRegularization
    activation_layer_output = None

    for layer in model.layers:
        if isinstance(layer, ActivationRegularization):
            activation_layer_output = layer.output
            break

    if activation_layer_output is None:
        print("Aucune couche ActivationRegularization trouvée.")
        return

    probe_model = Model(
        inputs=model.input,
        outputs=activation_layer_output
    )

    activations = probe_model.predict(X_sample, verbose=0)

    mean_per_neuron = activations.mean(axis=0)
    std_per_neuron = activations.std(axis=0)

    plt.figure(figsize=(12, 5))

    plt.subplot(1, 2, 1)
    plt.bar(range(len(mean_per_neuron)), mean_per_neuron)
    plt.axhline(RHO, color="red", linestyle="--", label=f"rho cible = {RHO}")
    plt.title("Taux d'activation moyen par neurone")
    plt.xlabel("Neurone")
    plt.ylabel("Activation moyenne")
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.bar(range(len(std_per_neuron)), std_per_neuron)
    plt.axhline(STD_MIN, color="red", linestyle="--", label=f"std_min = {STD_MIN}")
    plt.title("Ecart-type d'activation par neurone")
    plt.xlabel("Neurone")
    plt.ylabel("Ecart-type")
    plt.legend()

    plt.tight_layout()
    plt.show()


analyze_custom_activations(custom_model, X_test)


 
# SAUVEGARDE DES MODELES (OPTIONNEL)
 

baseline_model.save(
    "mlp_baseline.keras"
)

l2_model.save(
    "mlp_l2.keras"
)

custom_model.save(
    "mlp_distribution_regularization.keras"
)


print("\nEnregistrement terminé.")
```
